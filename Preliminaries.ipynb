## The course in a nutshell

**MCMC** is a sampling technique.

**Sampling** is drawing samples from a probability distribution.

**Probability distribution functions (PDF)** encode how likely it is for a random variable to assume a specific value/range of values.

<u>Example</u>. Normal (Gaussian) distribution with mean $\mu$ and dispersion $\sigma$:
$$
f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp{\left\{-\frac{(x-\mu)^2}{2\sigma^2}\right\}}
$$

Data can be fit to a model by sampling from the distribution of parameters of the models. MCMC comes in handy in doing that.
